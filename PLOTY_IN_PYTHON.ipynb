```python
import plotly.express as px
import pandas as pd
import matplotlib.pyplot as plt
from re import template
```

```python
data=px.data.gapminder()
data
```

```python
data=px.data.gapminder()
data.query('country == "India"')
```

```python
data=px.data.gapminder()
india_data = data.query('country == "India"')
fig = px.bar(india_data, x='year', y='pop', title='Population by India', labels={'pop':'Population'}, color='year', hover_data=['pop'], text='pop')
fig.update_traces(texttemplate='%{text:.2s}', textposition='inside')
fig.update_layout(uniformtext_minsize=8, uniformtext_mode='hide')
fig.update_layout(xaxis_title='year',yaxis_title='population',xaxis_tickangle=-45,yaxis_tickangle=-45,template="plotly_dark")
fig.show()
```

```python
data=px.data.gapminder()
india_data = data.query('country == "India"')
fig = px.line(india_data, x='year', y='pop', title='Population by India', labels={'pop':'Population'}, color='year', hover_data=['pop'], text='pop')
fig.update_traces( texttemplate='%{text:.2s}',textposition='top center')
fig.update_layout(uniformtext_minsize=8, uniformtext_mode='hide')
fig.update_layout(xaxis_title='year',yaxis_title='population',xaxis_tickangle=-45,yaxis_tickangle=-45,template="plotly_dark")
fig.show()
```

```python
data=px.data.gapminder()
india_data = data.query('country == "India"')
fig = px.scatter(india_data, x='gdpPercap', y='lifeExp', title='GDP per Capital in India', labels={'gdpPercap':'GDP per Capital'}, color='gdpPercap', hover_data=['gdpPercap'], text='gdpPercap')
fig.update_traces(texttemplate='%{text:.2s}', textposition='top center')
fig.update_layout(uniformtext_minsize=8, uniformtext_mode='hide')
fig.update_layout(xaxis_title='gdpPercap',yaxis_title='lifeExp',xaxis_tickangle=-45,yaxis_tickangle=-45,template="plotly_dark")
fig.show()
```

```python
data=px.data.gapminder()
india_data = data.query('country == "India"')
fig = px.pie(india_data,  values='gdpPercap', names='year', title='GDP per Capital in India', labels={'gdpPercap':'GDP per Capital'}, color='gdpPercap', hover_data=['gdpPercap'])
fig.update_traces( textposition='inside')
fig.update_layout(uniformtext_minsize=8, uniformtext_mode='hide')
fig.show()
```

```python
data=px.data.gapminder()
india_data = data.query('country == "India"')
fig = px.histogram(india_data, x='year', y='pop', title='Population by India', labels={'pop':'Population'}, color='year', hover_data=['pop'])
fig.update_traces( textposition='inside')
fig.update_layout(uniformtext_minsize=8, uniformtext_mode='hide')
fig.update_layout(xaxis_title='year',yaxis_title='population',xaxis_tickangle=-45,yaxis_tickangle=-45,template="plotly_dark")
fig.show()
```

```python
data=px.data.gapminder()
india_data = data.query('country == "India"')
fig = px.density_heatmap(india_data, x='year', y='pop', title='Population by India', labels={'pop':'Population'}, hover_data=['pop'])
fig.update_layout(uniformtext_minsize=8, uniformtext_mode='hide')
fig.update_layout(xaxis_title='year',yaxis_title='population',xaxis_tickangle=-45,yaxis_tickangle=-45,template="plotly_dark")
fig.show()
```

```python
data=px.data.gapminder()
india_data = data.query('country == "India"')
fig = px.box(india_data,y='lifeExp',points="all",color_discrete_sequence=["red"])
fig.update_traces(boxmean=False,jitter=0.6,Marker=dict(size=2,opacity=0.6))
fig.show()
```
